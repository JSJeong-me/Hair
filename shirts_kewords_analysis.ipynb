```python
# Combine all steps into a single Python program for the user

import pandas as pd
from collections import Counter
import re

def analyze_reviews(input_csv_path, output_excel_path, words_to_exclude):
    # Load the CSV file into a DataFrame
    df = pd.read_csv(input_csv_path)

    # Initialize a list to hold the filtered common words for each row
    filtered_common_words_per_row = []

    # Loop through each review in the 'review' column
    for review in df['review']:
        # Normalize and split the review into words
        words = re.findall(r'\b\w+\b', review.lower())
        # Count the frequency of each word in the review
        local_word_freq_counter = Counter(words)
        # Remove the specified words to exclude
        filtered_local_word_freq_counter = Counter({word: count for word, count in local_word_freq_counter.items() if word not in words_to_exclude})
        # Get the 10 most common words and their frequencies for this review
        local_common_words = filtered_local_word_freq_counter.most_common(10)
        # Append the filtered common words for this row to the list
        filtered_common_words_per_row.append(local_common_words)

    # Add the filtered common words for each row as a new column in the DataFrame
    df['Filtered_Common_Words'] = filtered_common_words_per_row

    # Save the updated DataFrame to a new Excel file
    df.to_excel(output_excel_path, index=False)

# Define input and output paths and words to exclude
input_csv_path = 'shirts.csv'
output_excel_path = 'shirts_with_filtered_common_words_program_output.xlsx'
words_to_exclude = ['너무', '잘', '것', '딱', '좀', '수', '살짝']

# Run the program
analyze_reviews(input_csv_path, output_excel_path, words_to_exclude)

output_excel_path
```
