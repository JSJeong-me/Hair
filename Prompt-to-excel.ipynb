```python
import os

# Define the directory where the uploaded files are stored
file_dir = ./'

# List of filenames to read
file_names = [f"AIL ({i}).txt" for i in range(1, 11)]

# Initialize an empty list to hold the lines from all files
all_lines = []

# Read each file and save the lines to the all_lines list
for file_name in file_names:
    file_path = os.path.join(file_dir, file_name)
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        all_lines.extend(lines)

# Show the first few lines for verification
all_lines[:10]
```

```python
import pandas as pd

# Create a DataFrame from the list of lines
df = pd.DataFrame(all_lines, columns=['Line'])

# Define the Excel file path to save the data
excel_file_path = 'lines_from_text_files.xlsx'

# Save the DataFrame to an Excel file
df.to_excel(excel_file_path, index=False)

excel_file_path
```
