```python
import pandas as pd

# 원본 문자열
original_string = 'a man with a tie and a white shirt, short hair cut, ail hair style, front face, smiling <lora:TEXTURE_LoRA-12:1>'

# 변경할 부분
replace_values = ['12:0.5', '12:0.6', '12:0.7', '12:0.8', '12:0.9', '12:1']

# 결과를 저장할 리스트
modified_strings = []

# 문자열을 수정
for value in replace_values:
    modified_string = original_string.replace('12:1', value)
    modified_strings.append(modified_string)

# 결과를 DataFrame으로 변환
df = pd.DataFrame(modified_strings, columns=['Modified String'])

# Excel 파일로 저장
excel_path = 'modified_strings.xlsx'
df.to_excel(excel_path, index=False)

excel_path
```
