```python
from PIL import Image
import os

def crop_upper_part(input_path, output_path):
    """
    Crop the upper part of an image and save it as a new file.

    Parameters:
    - input_path: str, path to the original image
    - output_path: str, path to save the cropped image
    """
    # Load the image
    image = Image.open(input_path)

    # Get image dimensions
    width, height = image.size

    # Crop the upper part of the image
    upper_part = image.crop((0, 0, width, height // 2))

    # Save the upper part as a new file
    upper_part.save(output_path)

def process_images_in_directory(directory_path):
    """
    Process all PNG images in the specified directory.

    Parameters:
    - directory_path: str, path to the directory containing images
    """
    for filename in os.listdir(directory_path):
        if filename.endswith('.png'):
            input_path = os.path.join(directory_path, filename)
            output_path = os.path.join(directory_path, 'upper_' + filename)
            crop_upper_part(input_path, output_path)
            print(f"Processed {filename}")


```
