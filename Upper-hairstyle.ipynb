```python
from PIL import Image

# 1. Load the image
input_path = 'GAIL.png'
image = Image.open(input_path)

# 2. Get image dimensions
width, height = image.size

# 3. Crop the upper part of the image
upper_part = image.crop((0, 0, width, height // 2))

# 4. Save the upper part as a new PNG file
output_path = 'GAIL-upper.png'
upper_part.save(output_path)

output_path
```
